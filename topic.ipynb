# Project 2. An HPLC calibration across gradient lengths

**Question.** Does the calibration slope of each analyte depend on the length of the solvent gradient, beyond the uncertainty from replicate injections, and does the scatter about each calibration line come from the injections or from the standards? (taken from CHEM 427 project list)

**Data.** data/knoevenagel.tar.bz2, committed as downloaded, an archive with one DAD1.CSV per HPLC run (calibration standards and blanks for 3 aldehydes, 5 gradient lengths, 2 replicate series). The CSVs are UTF-16 and are read from inside the archive with tarfile. In each CSV, one row is one time point: the unnamed first column is time in min (renamed time_min, 0.1 s steps), followed by absorbance at 351 wavelengths, columns 200 to 550 nm. Source: MOCCA example data, https://github.com/Bayer-Group/MOCCA/blob/example-data/knoevenagel.tar.bz2, accessed 2026-11-08 (Haas et al., ACS Cent. Sci. 9, 307 (2023)).

https://github.com/Bayer-Group/MOCCA/blob/example-data/knoevenagel.tar.bz2 (accessed 2026-11-08)



```python
import tarfile

import pandas as pd

archive_path = "data/knoevenagel.tar.bz2"  # MOCCA example data, as downloaded

# Read the first run's diode-array file straight from the archive; the CSVs are UTF-16
with tarfile.open(archive_path, "r:bz2") as archive:
    first_run = next(m for m in archive if m.name.endswith("DAD1.CSV"))
    data = pd.read_csv(archive.extractfile(first_run), encoding="utf-16")

data = data.rename(columns={"Unnamed: 0": "time_min"})  # time, min

print(first_run.name)
print(f"{len(data)} rows")
print(f"columns: {data.columns[0]}, then wavelengths {data.columns[1]} to {data.columns[-1]} nm ({len(data.columns) - 1} columns)")
print("absorbance: mAU")
```
